# Google BigQuery Data Wrangling: Hacker News Analytics
This notebook demonstrates clean data aggregation using the Google Cloud BigQuery API.

```python
import os
from google.cloud import bigquery
import pandas as pd

# Initialize Client
client = bigquery.Client()
safe_config = bigquery.QueryJobConfig(maximum_bytes_billed=10**10)

print("Setup Complete")
```

### 1) Prolific Commenters
Isolate accounts responsible for writing more than 10,000 records.

```python
prolific_commenters_query = """
SELECT `by` AS author, COUNT(1) AS NumPosts
FROM `bigquery-public-data.hacker_news.full`
GROUP BY `by`
HAVING COUNT(1) > 10000
"""

query_job_1 = client.query(prolific_commenters_query, job_config=safe_config)
prolific_commenters_df = query_job_1.to_dataframe()
prolific_commenters_df.head()
```

### 2) Deleted Comments Audit
Calculate the exact volume of records flagged with an active deletion status.

```python
deleted_comments_query = """
SELECT COUNT(1) AS num_deleted_comments
FROM `bigquery-public-data.hacker_news.full`
WHERE deleted = True
"""

query_job_2 = client.query(deleted_comments_query, job_config=safe_config)
deleted_comments_df = query_job_2.to_dataframe()
deleted_comments_df
```
